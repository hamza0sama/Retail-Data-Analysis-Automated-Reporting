### 📌 Question 6: Does offering higher discounts increase sales or reduce profit?


```python
df["Discount_Level"]=pd.cut(df["Discount"],
    labels=["No Discount","Low Discount","Medium Discount","High Discount","Very High Discount"],
    bins=[-0.01,0,0.2,0.5,0.7,1])
```

```python
Q6=df.groupby("Discount_Level",observed=True).agg(
    Total_Sales=("Sales","sum"),
    Average_Sales=("Sales","mean"),
    Total_Profit=("Profit","sum"),
    Average_Profit=("Profit","mean")
)
Q6
```

### 📌 Question 6: Does offering higher discounts increase sales or reduce profit?

#### 📊 Results

| Discount Level | Total Sales | Average Sales | Total Profit | Average Profit |
|---------------|------------:|--------------:|-------------:|---------------:|

| Low Discount | **846,522.18** | **222.59** | **100,785.47** | **26.50** |
| Medium Discount | **298,260.03** | **556.46** | **-58,804.95** | **-109.71** |
| High Discount | **47,264.98** | **85.01** | **-46,020.01** | **-82.77** |
| Very High Discount | **16,963.76** | **56.55** | **-30,539.04** | **-101.80** |

---

### 💡 Business Insights

- Orders without discounts generated the **highest total sales (1.09M)** and the **highest total profit (320.99K)**.
- **Low Discounts** remained profitable, although both total and average profit were lower than orders without discounts.
- **Medium Discounts** recorded the **highest average sales per order (556.46)**, indicating that customers tend to place higher-value orders when moderate discounts are offered.
- However, **Medium, High, and Very High Discounts all resulted in negative profits**, suggesting that the additional revenue was not enough to offset the cost of the discounts.
- As the discount level increases, profitability declines significantly, eventually turning negative.

---

### 📈 Key Findings

- Moderate discounts may encourage larger purchases.
- Excessive discounts have a strong negative impact on profitability.
- **No Discount** and **Low Discount** achieve the best balance between revenue and profit.
- **Medium to Very High Discounts** should be carefully evaluated before being applied.

> **Conclusion:** The analysis suggests that while moderate discounts may increase the average order value, higher discount levels consistently reduce profitability. The company should optimize its discount strategy to maximize both revenue and profit rather than relying on deep discounts.
